# Data Cleaning

```python
df = df.drop('CUST_ID', axis=1)
```

```python
min_payments_med = df['MINIMUM_PAYMENTS'].median()
df['MINIMUM_PAYMENTS'] = df['MINIMUM_PAYMENTS'].fillna(min_payments_med)
```

```python
cred_limit_med = df['CREDIT_LIMIT'].median()
df['CREDIT_LIMIT'] = df['CREDIT_LIMIT'].fillna(cred_limit_med)
```

```python
df.describe()
```

```python
df.describe().columns
```

# Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
df_feat = df[['BALANCE', 'BALANCE_FREQUENCY', 'PURCHASES', 'ONEOFF_PURCHASES',
       'INSTALLMENTS_PURCHASES', 'CASH_ADVANCE', 'PURCHASES_FREQUENCY',
       'ONEOFF_PURCHASES_FREQUENCY', 'PURCHASES_INSTALLMENTS_FREQUENCY',
       'CASH_ADVANCE_FREQUENCY', 'CASH_ADVANCE_TRX', 'PURCHASES_TRX',
       'CREDIT_LIMIT', 'PAYMENTS', 'MINIMUM_PAYMENTS', 'PRC_FULL_PAYMENT',
       'TENURE']]
```

```python
scaled_features = scaler.fit_transform(df_feat)
```

# PCA

```python
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=7)
```

```python
pca_feat = pca.fit_transform(scaled_features)
```

```python
print("Original Shape:", scaled_features.shape)
print("PCA Shape:", pca_feat.shape)
```

```python
pca_df = pd.DataFrame(data=pca_feat, columns=['PC1', 'PC2', 'PC3', 'PC4', 'PC5', 'PC6', 'PC7'])
```

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC1'],pca_df['PC2'], c=df['CREDIT_LIMIT'])
plt.xlabel('PC1')
plt.ylabel('PC2')
```

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC3'],pca_df['PC4'], c=df['CREDIT_LIMIT'])
plt.xlabel('PC3')
plt.ylabel('PC4')
```

# KMeans

```python
from sklearn.cluster import KMeans
```

```python
sse=[]
k_range=range(2,10)

for k in k_range:
    kmeans=KMeans(n_clusters=k)
    kmeans.fit(pca_df)
    sse.append(kmeans.inertia_)
```

```python
plt.plot(k_range, sse, color='blue', marker='o', markerfacecolor='red', markersize=10)
plt.xlabel('No. of clusters(k)')
plt.ylabel('SSE')
plt.title('Elbow Method')
```

Observation : Based on the graph, K=4 is a clear elbow point

```python
kmeans_final_model = KMeans(n_clusters=4, random_state=42)
```

```python
kmeans_final_model.fit(pca_df)
```

```python
kmeans_final_model.cluster_centers_
```

```python
cluster_labels = kmeans_final_model.labels_
```

```python
df['CLUSTER'] = cluster_labels
```

# Model Evaluation

```python
from sklearn.metrics import silhouette_score
```

```python
print('Silhouette Score:',silhouette_score(pca_df, df['CLUSTER']))
```

```python
sns.scatterplot(data=pca_df, x=pca_df['PC1'], y=pca_df['PC2'], c= df['CLUSTER'], cmap='rainbow')
```

# Cluster Profiling

```python
feats = ['BALANCE', 'PURCHASES','CASH_ADVANCE', 'CREDIT_LIMIT', 'PAYMENTS', 'PRC_FULL_PAYMENT','PURCHASES_FREQUENCY']
cluster_profile = df.groupby('CLUSTER')[feats].mean()
```

```python
print("Cluster Profiles: \n", cluster_profile.round(2))
```
